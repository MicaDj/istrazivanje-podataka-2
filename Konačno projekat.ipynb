```python
correlation_matrix_abs = X_reduced_var.corr().abs()   #matrica korelacija sa apsolutnim vrednostima
upper_triangle = correlation_matrix_abs.where(np.triu(np.ones(correlation_matrix_abs.shape), k=1).astype(bool))
pairs = upper_triangle.stack().loc[lambda x: x > 0.9].round(6)
pairs
```

```python
#Uklanjacemo iz svakog para korelisanih atributa onaj sa manjom varijansom, a ako je u paru jedan vec uklonjen onda drugi ostaje automatski
#Takodje ako su oba vec uklonjena, nastavljamo dalje
removed=set()
for a, b in pairs.index:   
    if a not in removed and b not in removed:
        if X_reduced_var[a].var() < X_reduced_var[b].var():
            removed.add(a)
        else:
            removed.add(b)
X_reduced_corr=X_reduced_var.drop(columns=removed)  
X_reduced_corr.to_csv("X_reduced_corr.csv", index=False)
print(len(removed))    
```

```python
#prvo gledamo na originalnim atributima: trazimo najbolje k na osnovu dve metode i predstavljamo graficki
wcss = []
silhouettes = []
K = range(2, 11)  #daje vrednosti od 2 do 10
for k in K:
    kmeans = KMeans(n_clusters=k, n_init=50, max_iter=500, init='k-means++', random_state=42)
    #n_init kaze koliko puta u jednoj iteraciji radi inicijalizacije i algoritam (na kraju izabere rezultat sa najmanjom inercijom)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)  #ciljna funkcija koju K-Means minimizuje, ukupna suma kvadrata rastojanja svih tačaka od njihovog centroida
    s = silhouette_score(X_scaled, kmeans.labels_)  #kmeans.labels_ je niz koji govori kom klasteru pripada svaka tačka nakon K-means klasterovanja.
    silhouettes.append(s)
    print(f'k={k}, WCSS={round(wcss[k-2], 2)}, Silueta koeficijent={round(silhouettes[k-2], 3)}')
```

```python
plt.plot(K, wcss, marker='o', color='#9FA1FF')
plt.xlabel('Broj klastera k')
plt.ylabel('WCSS')
plt.title('Metod lakta')
plt.savefig("km_metod_lakta_original.png", dpi=300, bbox_inches="tight")
plt.show()

markerline, stemlines, baseline = plt.stem(K, silhouettes)
plt.setp(markerline, color='#9FA1FF')
plt.setp(stemlines, color='#9FA1FF')
plt.setp(baseline, color='#9FA1FF')
plt.xlabel('Broj klastera k')
plt.ylabel('Silueta koeficijent')
plt.savefig("km_silueta_koef_original.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
#hijerarhijsko klasterovanje - aglomerativni pristup
#prvo cemo nacrtati dendrogram pomocu preuzete funkcije

# Funkcija za crtanje dendrograma
# preuzeta sa: https://scikit-learn.org/stable/auto_examples/cluster/plot_agglomerative_dendrogram.html
def plot_dendrogram(model, **kwargs):
    # Create linkage matrix and then plot the dendrogram
    # create the counts of samples under each node
    counts = np.zeros(model.children_.shape[0])
    n_samples = len(model.labels_)
    for i, merge in enumerate(model.children_):
        current_count = 0
        for child_idx in merge:
            if child_idx < n_samples:
                current_count += 1  # leaf node
            else:
                current_count += counts[child_idx - n_samples]
        counts[i] = current_count

    linkage_matrix = np.column_stack(
        [model.children_, model.distances_, counts]
    ).astype(float)
    # Plot the corresponding dendrogram
    dendrogram(linkage_matrix, **kwargs)
```
